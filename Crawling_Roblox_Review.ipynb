```python
from google_play_scraper import app, Sort, reviews_all
import pandas as pd
import numpy as np
```

```python
id_reviews = reviews_all(
    'com.roblox.client',
    sleep_milliseconds=0, # defaults to 0
    lang = 'id',
    country = 'id',
    sort = Sort.MOST_RELEVANT, # defaults to Sort.MOST_RELEVANT
    count = 15000
)
```

```python
print("Jumlah Data:", len(id_reviews))
```

```python
df = pd.DataFrame(id_reviews)
df.head()
```

```python
df.tail()
```

```python
df.to_csv('data_review_roblox.csv', index=False)
print("Data saved to 'data_review_roblox.csv'")
```
